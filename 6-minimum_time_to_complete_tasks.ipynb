# Question:

You are given an array consisting of several tasks, each with a different required time. You also have K identical workers available. Each worker takes a fixed amount of time to complete one unit of work.

Your task is to determine, given a time limit T, whether it is possible to complete all tasks within this time, under the following conditions:

* Each worker can only perform consecutive tasks. For example, if a worker is assigned tasks 1 and 2, then they cannot be assigned task 3 and then task 4; the tasks must be consecutive.

* A task cannot be divided between two workers.

# Answer:

### Optimal Greedy Approach:

1. Feasibility Check:

    * We are given an array of task durations and K workers.

    * Each worker can handle only consecutive tasks and must not exceed a given time limit T.

    * No task can be split across workers.

2. Greedy Partitioning Strategy:

    * Traverse the tasks array from left to right.

    * Use a variable current_time to track the total time assigned to the current worker.

    * If adding a task exceeds the limit T, assign the task to a new worker:

        - Reset current_time to the current task's time.

        - Increment the workers_used counter.

        - If workers_used > K, it's not possible to finish all tasks within time T.

    * Otherwise, add the task to the current worker's workload.


### Why This Works?

* The greedy strategy ensures each worker takes on as many consecutive tasks as possible without exceeding the allowed time T.

* By only moving to a new worker when necessary, we minimize the number of workers used.

* This directly matches the problem constraints and allows efficient feasibility checking.


### Time Complexity:

* O(n) – One linear pass through the tasks array.

```python
def assign_tasks(tasks, K, T):

    if max(tasks) > T:
        return False, []

    assignment = []
    current_worker = []
    current_time = 0
    workers_used = 1

    for task in tasks:
        if current_time + task > T:
            
            assignment.append(current_worker)
            current_worker = [task]
            current_time = task
            workers_used += 1

            if workers_used > K:
                return False, []

        else:
            current_worker.append(task)
            current_time += task

    assignment.append(current_worker)
    return True, assignment
```

```python
tasks = [3, 5, 2, 1, 4]
K = 2
T = 8

is_possible, assignment = assign_tasks(tasks, K, T)

if is_possible:
    print("Tasks can be completed within time T.")
    for i, worker_tasks in enumerate(assignment, 1):
        print(f"Worker {i}: {worker_tasks} (Total time: {sum(worker_tasks)})")
else:
    print("Not possible to assign tasks within time T.")
```
