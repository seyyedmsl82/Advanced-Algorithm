# Question:

You are given an array consist of the lengths of some ropes.

Your task is to connect all ropes into a single rope in such a way that the total cost of connections is minimized.

The cost of connecting two ropes is equal to the sum of their lengths.

Design an algorithm that minimizes this total connection cost.

# Answer:

### **Idea:**

To minimize the total cost, we use a Greedy Strategy using Min-Heap. We should always:

1. Pick the two shortest ropes first.

2. Combine them (paying the cost = their sum).

3. Add the new rope back into the pool.

4. Repeat until only one rope remains.


### Algorithm (Using Min-Heap / Priority Queue)

1. Insert all rope lengths into a min-heap.

2. While more than one rope remains:

    - Extract two smallest ropes x and y

    - Add cost x + y to total

    - Push x + y back into the heap

3. Final total cost is the answer.

```python
import heapq

def min_connection_cost(ropes):
    heapq.heapify(ropes)
    total_cost = 0

    while len(ropes) > 1:
        first = heapq.heappop(ropes)
        second = heapq.heappop(ropes)
        cost = first + second
        total_cost += cost
        heapq.heappush(ropes, cost)

    return total_cost

ropes = [4, 3, 2, 7]
print("Minimum total cost:", min_connection_cost(ropes))
```
