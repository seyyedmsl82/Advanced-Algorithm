# Question:

Iran will produce *m* barrels of oil this month. There is a list of *n* orders from potential buyers. In order *i*, the buyer is willing to purchase *aᵢ* barrels at a total price of *pᵢ* (this is the total price, not the price per barrel; Also, *pᵢ* can be negative)

### Sales Conditions:
* Each order must either be fully accepted or fully rejected. (Partial fulfillment is not allowed.)
* Each order can only be accepted once.
* We are not required to sell all of its oil. However, for each unsold barrel, we must pay a storage cost of *s* tomans.

Design an *O(nm)* time algorithm to determine which orders to accept in order to maximize profit.
(Note: The profit can be negative.)

# Answer:

### **We have:**

* m barrels of oil available for sale
* n orders: [(a₁, p₁), ..., (aₙ, pₙ)] where:
  - aᵢ = barrels required for order i
  - pᵢ = total price offered for order i (can be negative)
* Storage cost s per unsold barrel

### **Constraints:**
1. Each order must be fully accepted or rejected (no partial fulfillment)
2. Each order can only be accepted once
3. Unsold barrels incur storage cost s per barrel

### **Idea:**

We use a DP:

1. `dp[j] = maximum profit when having j barrels remaining`

2. **Initialization:** `dp[j] = -s * j` for all j (base case: no orders accepted, pay storage for all barrels)

3. For each order i from 1 to n:
   - For each remaining barrels j from m down to aᵢ:
     - If accepting order i gives better profit (`dp[j - aᵢ] + pᵢ > dp[j]`):
       - Update `dp[j] = dp[j - aᵢ] + pᵢ`
       - Mark `choice[i][j] = True`

4. The final answer is `dp[m]` and `accepted_orders` contains indices of accepted orders

5. **Backtracking:**
   - Start with `j = m` (all barrels initially)
   - For each order `i` from `n` down to 1:
     - If `choice[i][j]` is True:
       - Add order `i` to accepted list
       - Reduce remaining barrels: `j -= aᵢ`


### **Complexity Analysis:**

1. **Initialization:**
   - DP array: O(m)
   - Choice table: O(n × m)

2. **DP Filling:**
   - Outer loop: n iterations
   - Inner loop: m iterations (worst case)
   - Total: O(n × m)

3. **Backtracking:**
   - O(n) time to reconstruct solution

**Total Time Complexity:** O(n × m)  


### **Key Observations:**
1. The problem is a variant of the 0/1 Knapsack where:
   - "Capacity" = available barrels (m)
   - "Items" = orders with "weight" (aᵢ) and "value" (pᵢ)
   - "Storage cost" acts as a penalty for unused capacity

2. The reverse iteration (from m down to aᵢ) ensures each order is only used once

3. Negative prices are handled naturally by the DP approach

4. The solution automatically considers the trade-off between:
   - Accepting profitable orders
   - Rejecting unprofitable orders (even if that means paying storage costs)
   - Finding the optimal balance between these options

```python
from dataclasses import dataclass
from typing import List, Tuple

@dataclass
class Order:
    barrels: int
    price: int

def maximize_profit(m: int, s: int, orders: List[Order]) -> Tuple[int, List[int]]:
    n = len(orders)
    dp = [-s * j for j in range(m + 1)]
    choice = [[False] * (m + 1) for _ in range(n+1)]  # For backtracking

    for i in range(1, n + 1):
        a, p = orders[i-1].barrels, orders[i-1].price
        for j in range(m, a - 1, -1):
            if dp[j - a] + p > dp[j]:
                dp[j] = dp[j - a] + p
                choice[i][j] = True

    # Backtrack to find accepted orders
    accepted_orders = []
    j = m
    for i in range(n, 0, -1):
        if choice[i][j]:
            accepted_orders.append(i-1)
            j -= orders[i-1].barrels

    accepted_orders.reverse()
    return dp[m], accepted_orders

```

```python
orders = [
    Order(barrels=22, price=40),
    Order(barrels=11, price=70),
    Order(barrels=3, price=-10),
    Order(barrels=12, price=80),
]

m = 37
s = 5

profit, accepted = maximize_profit(m, s, orders)
print("Maximum Profit:", profit)
print("Accepted Orders:", accepted)
```
