# Question:

In a scientific competition, n participants are standing in a line. After the test, each participant is given a score. 

The organizers want to give each participant some number of books as a prize, under the following conditions:

* Each participant must receive at least one book.    
* Participants with higher scores must recieve more books than their neighbors.

Considering the scores of participants, determine the minimum total number of books that must be distributed. (using greedy)

# Answer:

### Optimal Greedy Approach:
1. Left-to-Right Pass:

    * Initialize an array books where each participant gets at least 1 book.

    * Traverse from left to right:

        - If the current participant's score is higher than their left neighbor, ensure they get more books than the left neighbor (`books[i] = books[i-1] + 1`).

2. Right-to-Left Pass:

    * Traverse from right to left:

        - If the current participant's score is higher than their right neighbor, ensure they get more books than the right neighbor (`books[i] = max(books[i], books[i+1] + 1)`).

3. Sum the Books:

    * The total minimum books required is the sum of the books array.

### Why This Works?

* The left-to-right pass ensures that higher-scored participants get more books than their left neighbors.

* The right-to-left pass ensures that higher-scored participants also get more books than their right neighbors.

* Combining both passes guarantees that both left and right constraints are satisfied with the minimal total books.



### Time Complexity:

* O(n) (two linear passes over the array).

* Space Complexity: O(n) (to store the books array).



```python
def min_books(scores):
    n = len(scores)
    if n == 0:
        return 0
    
    books = [1 for _ in range(n)]  # Everyone gets at least 1 book
    
    # Left to right pass
    for i in range(1, n):
        if scores[i] > scores[i-1]:
            books[i] = books[i-1] + 1
    
    # Right to left pass
    for i in range(n-2, -1, -1):
        if scores[i] > scores[i+1]:
            books[i] = max(books[i], books[i+1] + 1)
    print([f'{score, book}' for score, book in zip(scores, books)])
    return sum(books), [f'Score: {score} Number of Books: {book}' for score, book in zip(scores, books)]

print(min_books(scores = [1, 3, 2, 2, 1]))
print(min_books(scores = [4, 4, 3, 2, 1]))
print(min_books(scores = [5, 4, 3, 2, 2]))
```
