# Question

A sequence of n numbers (positive or negative) is given to you. Your task is to choose a subset of these numbers such that their sum is maximized, with the constraint that you cannot choose two consecutive numbers. That is, if you choose xₖ, you cannot choose xₖ₋₁ or xₖ₊₁.
Explain how, using dynamic programming with respect to n, you can find the subset that yields the maximum sum.

# Answer
We denote the array as arr[0..n-1]

Define a DP array:

        dp[i]: the maximum sum we can obtain considering elements from 0 to i, without selecting two adjacent elements.

Recurrence Relation:

        dp[0] = arr[0]
        dp[1] = max(arr[0], arr[1])
        For i ≥ 2:
        dp[i] = max(dp[i-1], dp[i-2] + arr[i])

The time complexity is O(n)

```python
def max_subset_finder(arr: list) -> int:
    length = len(arr)
    if length == 0:
        raise ValueError('The input list must have at least one element')
    
    elif length == 1:
        return arr[0]
    
    summation_matrix = [0 for _ in range(length)]
    summation_matrix[0] = arr[0]
    summation_matrix[1] = max(arr[0], arr[1])

    for i in range(2, length):
        summation_matrix[i] = max(summation_matrix[i-1], summation_matrix[i-2] + arr[i])

    return summation_matrix[-1]

arr = [0, 10, 22, 1, 32]
max_subset = max_subset_finder(arr)

print(max_subset)
```
